```python
def swishb(x):
    '''
    x：每个值
    '''
    return (K.sigmoid(x) * x / 2 - 0.5)
get_custom_objects().update({'swishb': Activation(swishb)})
```

```python
tf.compat.v1.disable_eager_execution()
tf.compat.v1.set_random_seed(sd)
def grad( y, x ):
    return Lambda( lambda z: K.gradients(z[0], z[1]), output_shape = [1])( [ y, x ] )
```

```python
class LnZ(keras.layers.Layer):
    '''return LnZ'''
    def __init__(self, deg=25, eta=0.0):
        '''use scaled Gauss Laguerre integration
        eta=0: for boltzmann gas
        eta=1: for fermion gas (quarks)
        eta=-1: for boson gas (gluond)'''
        super(LnZ, self).__init__()
        # fetch the nodes and weights for Gaussian integration
        xk, wk = np.polynomial.laguerre.laggauss(deg=deg)
        xk_ = keras.initializers.Constant(value = xk)
        wk_ = keras.initializers.Constant(value = wk)
        rk_ = keras.initializers.Constant(value = np.exp(-xk))
        
        self.eta = eta

        self.pnodes = self.add_weight(
            shape=(deg,),
            initializer=xk_,
            dtype="float32",
            trainable=False,
        )
        
        self.wnodes = self.add_weight(
            shape=(deg,),
            initializer=wk_,
            dtype="float32",
            trainable=False,
        )
        
        self.rnodes = self.add_weight(
            shape=(deg,),
            initializer=rk_,
            dtype="float32",
            trainable=False,
        )

    def call(self, inputs):
        T = inputs[0]
        m = inputs[1]
        scale = 1
        #scale = T # use T as the scale factor of gauss laguerre integration
        # problem: dlnZ/dT will depend on the numerical integration, does it matter
        psqure = tf.square(tf.multiply(self.pnodes, scale))
        E = tf.sqrt(psqure + tf.square(m))
        # use e^{-E/T} to avoid Nan in e^{E/T} when E/T is too large
        efactor = tf.exp(-tf.divide(E, T))
        f = tf.multiply(efactor, self.eta)
        # 1.0E-8 to avoid the BoseEinstein condensate and Nan numerically
        f = tf.add(f, 1+1.0E-8)        
        f = tf.math.log(f)
        f = tf.multiply(psqure*self.eta, f)
        f = tf.multiply(self.wnodes, f)
        f = tf.multiply(f, scale)
        f = tf.divide(f, self.rnodes)
        return tf.reduce_sum(f, axis=1)
```

```python
df = pd.read_csv("./data/hotqcd_1407.6387_err.csv")
Tarr = df['T']
SoverT3 = df['s/T^3']
EoverT4 = df['E/T^4']
PoverT4 = df['P/T^4']
```

```python
def residual_layer(inp, neurons=32, activation='swishb', reg=None):
    fc1 = Dense(neurons, activation=activation, kernel_regularizer=reg)(inp)
    fc2 = Dense(neurons, activation=activation, kernel_regularizer=reg)(fc1)
    return keras.layers.add([inp, fc2])

def mass_model(T, neurons=32, residual_layers=8, closer_test=False, activation='swishb'):
    reg = None
    mass1 = Dense(neurons, activation=activation, kernel_regularizer=reg)(T)
    for n in range(residual_layers):
        mass1 = residual_layer(mass1, neurons, activation=activation, reg=reg)
    mass1 = Dense(1, activation='sigmoid')(mass1)

    mass2 = Dense(neurons, activation=activation, kernel_regularizer=reg)(T)
    for n in range(residual_layers):
        mass2 = residual_layer(mass2, neurons, activation=activation)        
    mass2 = Dense(1, activation='sigmoid')(mass2)
    M = keras.layers.add([mass1, mass2])

    M = 3 * M
    return M

def SU2p1_lnZ_model(T, neurons=32, residual_layers=8, closer_test=False, activation='swishb'):
    ud_mass = mass_model(T, neurons=32, residual_layers=8, activation='swishb')
    s_mass = mass_model(T, neurons=32, residual_layers=8, activation='swishb')
    
    gluon_mass = mass_model(T, neurons=32, residual_layers=8, activation='swishb')
    
    ud_dof = 2 * 2 * 2 * 3  #u,d * q, qbar * spin_dof * color_dof
    s_dof = 2 * 2 * 3
#     c_dof = 2 *2 * 3
    gluon_dof = 8 * 2       # color dof * polarization dof
    ud_lnZ = ud_dof * LnZ(eta=1)([T, ud_mass])
    s_lnZ = s_dof * LnZ(eta=1)([T, s_mass])
#     c_lnZ = c_dof * LnZ(eta=1)([T, c_mass])
    g_lnZ =  gluon_dof * LnZ(eta=-1)([T, gluon_mass])
    lnZ = keras.layers.add([ud_lnZ, s_lnZ, g_lnZ])
    
    return lnZ, ud_mass, s_mass, gluon_mass
```

```python
T = keras.layers.Input(1)
    
lnZ, Mud, Ms, Mg = SU2p1_lnZ_model(T, neurons=32, closer_test=False)


coef =  1 / (2 * np.pi**2)

ed = coef * T * T * grad(lnZ, T) 

pr = coef * T * lnZ

mass_loss_1 = tf.where(T >(2.5 * 0.150), 0.01*tf.abs(Mg/Mud - 1.5), 0)
mass_loss_2 = tf.where(T >(2.5 * 0.150), 0.0006*tf.abs((Ms  - Mud)/0.09 - 1), 0)

L_mass = mass_loss_1 + mass_loss_2




s = (ed + pr) / T 


Delta = (ed - 3 * pr)

model = keras.Model(T, Delta)

m_model = keras.Model(T, [Mud, Ms,Mg])

e_model = keras.Model(T, ed)
p_model = keras.Model(T, pr)
s_model = keras.Model(T, s)

quasi_model = keras.Model(T, [s, Delta, L_mass])
```

```python
quasi_model.compile(loss='mse', optimizer=keras.optimizers.Adam(lr=0.0001))
```

```python
class EarlyStoppingByLossVal(keras.callbacks.Callback):
    def __init__(self, monitor='loss', value=0.00001, verbose=0):
        super(keras.callbacks.Callback, self).__init__()
        self.monitor = monitor
        self.value = value
        self.verbose = verbose

    def on_epoch_end(self, epoch, logs={}):
        current = logs.get(self.monitor)
        if current is None:
            warnings.warn("Early stopping requires %s available!" % self.monitor, RuntimeWarning)

        if current < self.value:
            if self.verbose > 0:
                print("Epoch %05d: early stopping THR" % epoch)
            self.model.stop_training = True
```

```python
class PlotCallback(keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs={}):
        plot_trace_anomally(epoch)
        plot_ed(epoch)
        plot_pr(epoch)
        plot_entropy(epoch)
        plot_mass(epoch)
        data_epoch(epoch)
        return

visualize = PlotCallback()

callbacks_list = [
        visualize,
        keras.callbacks.ReduceLROnPlateau(monitor='loss', factor=0.9, patience=1000)
        ]
```

```python

Sarr = SoverT3 * Tarr**3
Earr = EoverT4 * Tarr**4
Parr = PoverT4 * Tarr**4
TraceAnomaly = Earr - 3 * Parr
```

```python
epochs = 100000

quasi_model.fit(Tarr, [Sarr, TraceAnomaly, np.zeros(len(Tarr))],
            epochs=epochs,
            batch_size=1,
            shuffle=False,
            callbacks=callbacks_list)
```
